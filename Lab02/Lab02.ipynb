Task 06:

```python
provinces = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
census17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop_23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def D(two_rolls):
    return {str(k): (val.item() if hasattr(val, "item") else val) for k, val in two_rolls}

total_pop = pop_23.sum()
print(f"T6a: total population 2023 = {total_pop:.2f} million")

share_pct = 100 * pop_23 / total_pop
print(f"T6b: share of total (%): {D(zip(provinces, np.round(share_pct, 2)))}")

print(f"T6c: most populous unit: {provinces[np.argmax(pop_23)]}")
print(f"T6d: units above 20 million: {provinces[np.where(pop_23 > 20)]}")

cagr = (np.power(pop_23 / census17, 1 / 6) - 1) * 100
print(f"T6e: annual growth 2017-2023 (% per year): {D(zip(provinces, np.round(cagr, 2)))}")

print(f"T6f: fastest-growing unit: {provinces[np.argmax(cagr)]} ({cagr.max():.2f}% per year)")

national_growth = ((pop_23.sum() / census17.sum()) ** (1 / 6) - 1) * 100
doubling = 70 / cagr.max()
print(f"T6 interpret: national growth = {national_growth:.2f}% per year.")
print(f"T6 interpret: {provinces[np.argmax(cagr)]} would double in about {doubling:.0f} years at its current rate.")
print(f"T6 interpret: AJK grew {cagr[-1]:.2f}% per year vs national {national_growth:.2f}% -> "
      f"AJK is {national_growth - cagr[-1]:.2f} points below the national rate (in this sample AJK is basically flat).")
```

Task 07:

```python
city_names = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
month_arr = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temp_data = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rain_grid = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def D(two_rolls):
    return {str(k): (val.item() if hasattr(val, "item") else val) for k, val in two_rolls}

print(f"T7a: shape of temp (cities, months): {temp_data.shape}")

yearly_mean = temp_data.mean(axis=1)
print(f"T7b: yearly mean temperature (deg C): {D(zip(city_names, np.round(yearly_mean, 1)))}")

hot_month = month_arr[temp_data.argmax(axis=1)]
print(f"T7c: hottest month per city: {D(zip(city_names, hot_month))}")

print(f"T7d: 5-city monthly average (deg C): {D(zip(month_arr, np.round(temp_data.mean(axis=0), 1)))}")

hot_months = np.count_nonzero(temp_data > 35, axis=1)
print(f"T7e: months above 35 deg C: {D(zip(city_names, hot_months))}")

lahore_deg_f = temp_data[0] * 1.8 + 32
print(f"T7f: Lahore monthly temperature (deg F): {np.round(lahore_deg_f, 1)}")

monsoon_mm = rain_grid[:, 6:9].sum(1)
print(f"T7g: monsoon rain Jul-Sep (mm): {D(zip(city_names, monsoon_mm))}")
print(f"T7g: wettest monsoon city: {city_names[np.argmax(monsoon_mm)]} ({monsoon_mm.max()} mm)")

share_pct = 100 * monsoon_mm / rain_grid.sum(axis=1)
print(f"T7h: monsoon share of annual rain (%): {D(zip(city_names, np.round(share_pct, 1)))}")

k = np.where(city_names == "Karachi")[0][0]
print(f"T7 interpret: Karachi has the highest yearly mean ({yearly_mean.max():.1f} deg C) "
      f"but {hot_months[k]} months above 35 deg C, while cooler-on-average Peshawar has {hot_months[4]} "
      f"and Lahore {hot_months[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

Task 08:

```python
fy_names = np.array(["FY2023", "FY2024", "FY2025"])
fm = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

inflation_grid = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def D(two_rolls):
    return {str(k): (val.item() if hasattr(val, "item") else val) for k, val in two_rolls}

avg = inflation_grid.mean(axis=1)
print(f"T8a: average inflation per year (%): {D(zip(fy_names, np.round(avg, 2)))}")

infl24 = inflation_grid[1]
print(f"T8b: FY2024 months above 25 %: {fm[infl24 > 25]} -> {np.sum(infl24 > 25)} months")

mom_change = inflation_grid[:, 1:] - inflation_grid[:, :-1]
print(f"T8c: month-on-month change (percentage points; negative = drop):\n{np.round(mom_change, 1)}")

max_drop_idx = np.argmin(mom_change, axis=1)
print(f"T8d: month of biggest drop: {D(zip(fy_names, fm[max_drop_idx + 1]))} | size (pp): {np.round(mom_change.min(axis=1), 1)}")

fy25_row = inflation_grid[2]
label = np.where(fy25_row < 5, "Low", np.where(fy25_row < 10, "Moderate", "High"))
print(f"T8e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fm, fy25_row, label))}")

print(f"T8 interpret: average inflation went {avg[0]:.1f}% -> {avg[1]:.1f}% -> {avg[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

Task 09:

```python
scores = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rolls20 = np.arange(1, 21)
PASSING = 40

marks_g = np.clip(scores + 5, 0, 100)
print(f"T9a: original marks: {scores}")
print(f"T9a: marks after +5 grace (cap 100): {marks_g}")

before = np.mean(scores >= PASSING) * 100
pass_after = 100 * np.mean(marks_g >= PASSING)
print(f"T9b: pass % before grace = {before:.1f} %, after grace = {pass_after:.1f} %")

saved_by_grace = np.logical_and(scores < PASSING, marks_g >= PASSING)
print(f"T9c: passed only because of grace (roll numbers): {rolls20[saved_by_grace]}")
print(f"T9c: their original marks: {scores[saved_by_grace]}")

z_score = (marks_g - marks_g.mean()) / marks_g.std()
print(f"T9d: mean = {marks_g.mean():.2f}, std = {marks_g.std():.2f}")
print(f"T9d: z-scores (std devs from mean): {np.round(z_score, 2)}")
```

```python
bitmap = np.zeros((8, 8), dtype=np.uint8)
bitmap[1, 1:7] = 255
for r, c in zip([2, 3, 4, 5, 6], [6, 5, 5, 4, 4]):
    bitmap[r, c] = 255

show = lambda a: (a > 0).astype(int)

print(f"T10a: the '7' (1 = white):\n{show(bitmap)}")

mirrored = bitmap[:, ::-1]
print(f"T10b: flipped left-right:\n{show(mirrored)}")

img_inv = 255 - bitmap
print(f"T10c: inverted (white <-> black):\n{show(img_inv)}")

cropped_img = bitmap[:7, 1:]
print(f"T10d: cropped 7x7 region, shape: {cropped_img.shape}\n{show(cropped_img)}")

bright_img = np.clip(bitmap.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"T10e: brightened (+60, capped at 255); unique values: {np.unique(bright_img)}")

print(f"T10f: shape = {bitmap.shape} | dtype = {bitmap.dtype} | white pixels = {(bitmap == 255).sum()}")
```
